```python
# Célula 1: Importações
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# Célula 2: Carregamento dos Dados (com limpeza de valores nulos)
df = pd.read_csv('creditcard.csv')

# Remove linhas onde a coluna Class é nula (evita o erro do train_test_split)
df = df.dropna(subset=['Class'])

print("Dimensões:", df.shape)
print("Proporção de fraudes:")
print(df['Class'].value_counts(normalize=True) * 100)

# Célula 3: Preparação e Engenharia de Atributos
# Transformando o Amount com Log para reduzir assimetria
df['LogAmount'] = np.log(df['Amount'] + 1)

X = df.drop(columns=['Class', 'Amount'])
y = df['Class']

# Padronizando variáveis
scaler = StandardScaler()
X['LogAmount'] = scaler.fit_transform(X[['LogAmount']])
X['Time'] = scaler.fit_transform(X[['Time']])

# Divisão treino/teste estratificada (essencial para manter a proporção de fraudes)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Célula 4: Treinamento do Modelo (Baseline com Regressão Logística Balanceada)
model = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
model.fit(X_train, y_train)

# Célula 5: Avaliação de Desempenho (Foco em Recall e F1)
y_pred = model.predict(X_test)

print("Matriz de Confusão:")
print(confusion_matrix(y_test, y_pred))

print("\nRelatório de Classificação:")
print(classification_report(y_test, y_pred))
```
